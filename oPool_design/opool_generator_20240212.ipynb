```python
###Import barcode file information###

bc_pdr1 = pd.read_csv('/Users/alexandre/Desktop/Projets/2023/IsabelleGArsenault/barcode_pdr1_fin.csv', sep='\t', index_col=0)
bc_pdr1.head()
```

```python
###Generate all opool by combining the different piece together and replacing the codon by NNK###

df_opool = pd.DataFrame(columns=['oPool', 'NNK_seq', 'to_order'])

for i in df_seq_split.index:
    
    DNA = df_seq_split.loc[i]['dms_part']
    DNA_length = len(df_seq_split.loc[i]['dms_part'])
    number_of_codons = int(DNA_length / 3)

    codon_list = []
    
    for j in range(number_of_codons) :
        
        codon_list.append(DNA[j *3 : j * 3 +3])
    
    hld1 = df_seq_split.loc[i]['aa_position'].split('-')[0]
    hld2 = df_seq_split.loc[i]['aa_position'].split('-')[1]
    
    hld3 = tuple(range(int(hld1), int(hld2)+1))
    
    for k in range(number_of_codons):
        
        codon_list2 = codon_list.copy()
        
        codon_list2[k] = 'NNK'

        seq = ''.join(codon_list2)
        
        df_opool.at[hld3[k], 'NNK_seq'] = seq
        df_opool.at[hld3[k], 'oPool'] = int(i+1)
        df_opool.at[hld3[k], 'to_order'] = df_seq_split.loc[i]['homo5prime'] + df_opool.loc[hld3[k]]['NNK_seq'] + df_seq_split.loc[i]['homo3prime'] + df_seq_split.loc[i]['bsaI_mod'] + bc_pdr1.loc[hld3[k]]['seq'] + df_seq_split.loc[i]['pbsi7']
    
    
    
        
```
